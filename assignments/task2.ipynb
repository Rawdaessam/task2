** Question 2
 Calculate Circle or Rectangle Area
**

```python

def area(shape, x, y=0):
    if shape.lower() == "circle":
        print(3.14 * x * x)

    elif shape.lower() == "rectangle":
        print(x * y)

    else:
        print("Invalid shape")


shape = input("Enter circle or rectangle: ").lower()

if shape == "circle":
    x = float(input("Enter radius: "))
    area(shape, x)

elif shape == "rectangle":
    x = float(input("Enter length: "))
    y = float(input("Enter width: "))
    area(shape, x, y)

else:
    print("Invalid shape")
```

** Question 5
 Restaurant Class**






```python

class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer_name, table_number):
        reservation = {
            "customer": customer_name,
            "table": table_number
        }
        self.book_table.append(reservation)

    def customer_order(self, customer_name, item):
        if item in self.menu_items:
            order = {
                "customer": customer_name,
                "item": item,
                "price": self.menu_items[item]
            }
            self.customer_orders.append(order)
        else:
            print("Item not found in menu!")

    def print_menu(self):
        print("Menu:")
        for item, price in self.menu_items.items():
            print(item, "-", price)

    def print_table_reservations(self):
        print("Table Reservations:")
        for reservation in self.book_table:
            print(reservation)

    def print_customer_orders(self):
        print("Customer Orders:")
        for order in self.customer_orders:
            print(order)




restaurant = Restaurant()

restaurant.add_item_to_menu("Pizza", 150)
restaurant.add_item_to_menu("Burger", 100)

restaurant.book_tables("Ahmed", 1)
restaurant.customer_order("Ahmed", "Pizza")

restaurant.print_menu()
restaurant.print_table_reservations()
restaurant.print_customer_orders()
```

 Question 6
Bike Rental System


```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, new_name):
        self.name = new_name

    def get_age(self):
        return self.age

    def set_age(self, new_age):
        self.age = new_age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, new_name):
        self.name = new_name

    def get_price(self):
        return self.price

    def set_price(self, new_price):
        self.price = new_price


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, new_bikes):
        self.bikes = new_bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, new_rentals):
        self.rentals = new_rentals

    def addBike(self, bike):
        self.bikes.append(bike)

    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6 years")
            return

        if number_of_bikes <= 0:
            print("Invalid number of bikes")
            return

        if number_of_bikes <= len(self.bikes):
            rented_bikes = []

            for i in range(number_of_bikes):
                rented_bikes.append(self.bikes.pop())

            rental = {
                "customer": customer.get_name(),
                "bikes": rented_bikes
            }

            self.rentals.append(rental)
            print("Bikes rented successfully")

        else:
            print("Wait 10 mins")

    def returnBike(self, rentalTime, num_of_bikes):
        if num_of_bikes <= 0 or rentalTime < 0:
            print("Invalid rental details")
            return 0

        returned_count = 0

        # Return bikes from the most recent rentals first
        while returned_count < num_of_bikes and self.rentals:
            rental = self.rentals[-1]

            while (
                rental["bikes"]
                and returned_count < num_of_bikes
            ):
                bike = rental["bikes"].pop()
                self.bikes.append(bike)
                returned_count += 1

            if not rental["bikes"]:
                self.rentals.pop()

        if returned_count < num_of_bikes:
            print("Not enough rented bikes to return")
            return 0

        total_bill = rentalTime * 40
        print("Total bill:", total_bill)

        return total_bill

    def totalCost(self):
        total = 0

        for bike in self.bikes:
            total += bike.get_price()

        return total




customer1 = Customer("Ahmed", 20)

bike1 = Bike("Mountain Bike", 5000)
bike2 = Bike("City Bike", 3000)
bike3 = Bike("Road Bike", 4000)

shop = MainBikeRental()

shop.addBike(bike1)
shop.addBike(bike2)
shop.addBike(bike3)

print("Available bikes:", len(shop.get_bikes()))

shop.requestBike(customer1, 2)

print("Available bikes after rental:", len(shop.get_bikes()))

shop.returnBike(3, 2)

print("Available bikes after return:", len(shop.get_bikes()))
print("Total cost of bikes in store:", shop.totalCost())
```
